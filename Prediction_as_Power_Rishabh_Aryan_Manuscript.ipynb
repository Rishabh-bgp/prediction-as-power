## Abstract

European wholesale electricity is no longer priced primarily by the slow economics of fuel. It is priced by expectations of weather. Wind, as a near-zero short-run marginal-cost resource, occupies the base of the merit order. The day-ahead auction therefore clears expected residual load rather than realised residual load. This paper treats prediction itself as a form of market power: not the power to withhold physical capacity, but the power to occupy the information set on which sequential markets are cleared. A structurally calibrated two-year hourly laboratory of a Germany-like bidding zone and a hydro-rich neighbour is used to compare a legacy numerical weather prediction (NWP) wind forecast with an artificial-intelligence (AI) forecast whose day-ahead mean absolute error is 34.2 percent lower. Substituting the AI wind input inside an otherwise unchanged electricity-price model shifts the reconstructed day-ahead price by 12.90 EUR/MWh on average in absolute terms, with a ninetieth-percentile absolute shift of 30.89 EUR/MWh. For a one-hundred-megawatt merchant slice, annualised imbalance cost falls from 6.19 million EUR under the NWP regime to 2.86 million EUR under the AI regime. A stylised combined-cycle commitment rule records 1,224 false thermal commitments under NWP residual forecasts against 970 under AI. Cross-border export to the hydro neighbour falls as expected home-zone wind rises. The results are laboratory magnitudes, not official ENTSO-E point estimates. They nevertheless make operational the claim that forecast quality reorders which plants run, which interconnectors flow, and who captures the rent to information in coupled European markets.


## Keywords

Artificial intelligence, wind power forecasting, European electricity markets, merit-order effect, imbalance costs, day-ahead coupling, Dunkelflaute, information asymmetry, residual load.


## I. INTRODUCTION

The European power system has completed a quiet inversion. For most of the twentieth century the short-run price of electricity was a fuel price passed through a heat rate, plus a scarcity rent when plant was tight. Demand moved slowly. Supply was dispatchable. Weather entered the problem mainly as heating load. That world has ended. Variable renewable energy now supplies a first-order share of generation in Germany, Denmark, Spain, Ireland and growing shares elsewhere. Wind and solar enter the supply stack at near-zero short-run marginal cost. The residual load that conventional plant, storage and interconnectors must serve is the difference between demand and a weather-determined injection. Because Single Day-Ahead Coupling closes around noon on the day before delivery, the quantity that is actually priced is not realised wind. It is a forecast of wind. Single Intraday Coupling is a market in subsequent revisions of that forecast. Balancing settles whatever remains after the last gate.

This institutional sequence converts meteorology into cash-flow and cash-flow into dispatch. A wind operator who sold two hundred megawatts in the day-ahead auction and later expects one hundred and fifty must buy fifty megawatts in the intraday market. If the revision arrives after the last gate, the fifty megawatts become an imbalance. The imbalance price is typically more extreme than the day-ahead price and is itself a function of the system-wide forecast error. Forecast error is therefore not a statistical residual. It is a priced object. At the same time, because wind sits at the foot of the merit order, a revision of expected wind by a few gigawatts moves the intersection of supply and demand along a steep segment of residual supply: from lignite onto gas, from gas onto scarcity pricing, or from a modestly positive price into a negative-price hour. The derivative of price with respect to expected wind is state-dependent and convex in tightness. The same gigawatt of error is cheap in a well-supplied afternoon and expensive in a winter Dunkelflaute.

Classical industrial organisation treats market power as the ability to move price by withholding quantity. In a high-renewable electricity system the scarce object is not quantity in the same physical sense. Wind and solar arrive as weather. What is scarce is a usable description of that weather at the moment each sequential market closes. An agent who systematically occupies a finer filtration of the weather sigma-algebra extracts a rent that looks, in the accounts, like avoided imbalance, and looks, in the power system, like the right to say which machines run. That is the sense in which this paper speaks of prediction as power. The claim is institutional before it is statistical. It is not a claim that artificial intelligence creates energy. It is a claim that artificial intelligence reallocates the right to describe energy that weather will create.

Artificial intelligence changes the distribution of that description. Physics-based numerical weather prediction systems such as the European Centre for Medium-Range Weather Forecasts Integrated Forecasting System produce a small number of official runs per day on large high-performance computers. AI systems, including the Centre’s own Artificial Intelligence Forecasting System, GraphCast-class models and Google WeatherNext-class models, update more frequently, assimilate historical error structure, and have been reported in industry evaluations to reduce one-hundred-metre wind mean absolute error by the order of twenty-five to thirty percent relative to the IFS control run. The same evaluations have translated those meteorological gains into imbalance-cost reductions of several hundred thousand euros per one hundred megawatts of wind over a half-year window. Those figures are commercial. They are also political. They describe a change in who can occupy the residual-load forecast that the rest of the market must take as given.

Two clarifications keep the claim from over-reaching. First, this is not, in the competition-law sense, a claim of abuse of dominance. No capacity is withheld. No offer curve is physically removed from the auction. The asymmetry is informational. It is closer to Hayek’s problem of the use of knowledge in society than to the classical monopolist’s problem, except that the knowledge is now produced by models whose weights are proprietary. Second, European market design already attempts to socialise a baseline description of the system. Regulation 543/2013 and the ENTSO-E Transparency Platform publish load forecasts, wind and solar forecasts, actual generation by type, and day-ahead prices. Transmission system operators publish their own renewable forecasts. The public baseline is real. What AI privatises is the residual around that baseline. The equilibrium between those two forces is the live regulatory question.

The merit-order effect of realised wind and solar on European prices is well documented. A related literature shows that forecast revisions, not only realisations, move German and coupled intraday prices and induce cross-border re-dispatch. Deep-learning electricity price forecasting across bidding zones has improved point-forecast accuracy and has complicated the use of volatility as a proxy for unpredictability. Probabilistic forecasts of load and renewable injection improve day-ahead price models, with extreme quantiles often more informative than means. What remains under-articulated is the political-economic content of these statistical gains. Forecast quality is usually treated as a technical input to an otherwise unchanged market. This paper treats forecast quality as a constitutive input. The market that clears under one information set is not the same market that clears under another.

The contribution is fourfold. First, the paper states a compact institutional mapping from sequential European markets to a filtration of weather information, and from that filtration to residual-load pricing with a scarcity term. Second, it constructs a structurally calibrated two-year hourly laboratory of a Germany-like home zone and a hydro-rich neighbour, with an explicit fleet power curve, persistent forecast errors, a convex price function and a stylised interconnector. Third, it implements a suite of experiments that make the thesis operational: a lead-time error ladder, reduced-form merit-order and value-factor diagnostics, tree-based electricity price forecasting with a counterfactual AI wind input, quantile wind forecasts, merchant imbalance cash-flow for a one-hundred-megawatt slice, a stylised combined-cycle commitment rule, a supervised power model sitting on the NWP run, and a cross-border wind shock. Fourth, it interprets the magnitudes as an ordering of agents—producers, thermal owners, hydro and interconnectors, system operators, consumers, forecast vendors and regulators—and as a design problem rather than a purely statistical one.

The laboratory is not a substitute for an ENTSO-E estimation. Public Transparency Platform feeds require a registered security token. The generator is therefore synthetic, but it is structural rather than black-box: every series can be inspected, and every coefficient can be replaced by an official extract without rewriting the economic layer. Coefficients in the results section are orders of magnitude. They are offered as a proof of mechanism, not as official statistics for a named year.

The remainder of the paper is organised as follows. Section II reviews the related literature on merit-order effects, forecast-driven trading, electricity price forecasting, AI weather models and the political economy of information in infrastructure markets. Section III describes the institutional architecture of coupled European electricity markets in more detail. Section IV states the theoretical framework. Section V presents the laboratory and calibration. Section VI describes the empirical design of each experiment. Section VII reports results. Section VIII discusses distribution, regulation and market design. Section IX states limitations and a path to a non-synthetic sequel. Section X concludes.


## II. RELATED LITERATURE


### 2.1 The merit-order effect and the value of wind

The merit-order effect is the observation that an additional megawatt-hour of near-zero marginal-cost generation displaces the most expensive accepted conventional unit and thereby lowers the uniform clearing price. Early German simulation and econometric work established that wind and solar feed-in dampen spot prices, with estimated effects on the order of a fraction of a euro per megawatt-hour per additional gigawatt-hour in mid-2010s conditions. Subsequent work showed that the slope is state-dependent: it flattens when fuel and carbon prices fall and steepens when the system is tight. Negative prices appear when must-run generation plus high variable renewable output exceeds demand net of export. These facts are now textbook. They are also incomplete. They describe what realised wind does to the stack. They do not describe what expected wind does to the stack at noon on $D{-}1$, which is the moment at which slow thermal plant is committed and at which most volume is still traded.

A parallel literature measures the value factor of wind and solar: the generation-weighted captured price divided by the time-weighted average price. A value factor below one is cannibalisation. Wind and solar produce most when they, and often each other, are already depressing the price. Solar faces a stronger cannibalisation risk than wind in many European zones because its production is more concentrated in the same midday hours. Value factors declined in most bidding zones between the mid-2010s and the mid-2020s as installed variable renewable capacity grew. Forecast quality does not repeal cannibalisation. It changes who captures the remaining scarce hours and who pays to clean up the hours that were mis-described at the day-ahead gate.


### 2.2 Forecasts as trading objects

Once wind is large enough to move the stack, wind forecasts become trading objects. Producers, utilities and financial traders bid day-ahead volumes that are forecasts. They then trade revisions in the continuous intraday market and in discrete intraday auctions. Empirical work on Germany has shown that wind and solar forecasts affect both day-ahead and intraday prices, and that reductions in forecast error dampen price volatility. Coupled-market studies have shown that German wind forecast updates are not balanced in Germany alone. Adjacent zones and a mix of technologies absorb the revision. Cross-border transmission is a shock absorber for information as much as for energy. That observation is central to this paper. A proprietary improvement in a North Sea wind forecast is an input to Norwegian hydro release, French nuclear ramps and Alpine pumped storage, subject to available transfer capacity.

The balancing market closes the sequence. Imbalance prices depend on the sign and size of the system imbalance, which is itself often the sum of renewable forecast errors and load forecast errors. A producer whose own error has the same sign as the system error is on the wrong side of a stressed price. A producer whose error offsets the system is paid for the coincidence. Accurate prediction therefore has two cash-flow channels: it reduces the volume of imbalance, and it reduces the covariance of that volume with system stress. Industry evaluations of AI weather models have begun to price the first channel. The second channel is less often reported and, in a convex system, may be the larger one in the tail.


### 2.3 Electricity price forecasting

Electricity price forecasting has become a large applied field. Expert models with lagged prices, load, wind, solar, calendar effects and fuel prices remain strong baselines. Regularised high-dimensional regressions and deep-learning architectures have been shown to improve day-ahead accuracy across many European bidding zones. Recent work covering nineteen zones in eleven countries from 2018 to 2023 found that deep-learning price forecasts weaken the link between volatility and unpredictability: a volatile market is not necessarily an unforecastable one once the right function class is used. Other work has shown that meteorological forecasts of wind speed, direction, irradiation, cloud cover and temperature improve German wholesale price forecasts at horizons of two to four days, with northern German wind speed a particularly strong explanatory variable. Feature-importance studies spanning 2020 to 2024 have documented the rising role of onshore wind and gas-fired generation as price drivers as capacity mixes have changed.

A further development is the use of probabilistic forecasts of fundamentals as features in price models. Point forecasts of load and renewable generation discard the tails. In a convex supply curve the tails are where prices move. Studies on the German EPEX market have reported root-mean-square-error reductions of up to about thirteen percent when quantile forecasts of load, wind and solar replace or complement point forecasts, with extreme quantiles among the most influential predictors. This paper takes that lesson into the laboratory by estimating a small quantile wind model and by insisting that mean absolute error is not a sufficient statistic for the economic value of a forecast.


### 2.4 Artificial intelligence in weather and power prediction

The last three years have produced a first generation of operational AI weather models from major forecasting centres and technology firms. Unlike classical NWP, which integrates discretised physical equations forward from an analysed initial state, these models learn the evolution of the atmosphere from historical reanalysis and observations. They are cheaper to run, faster to update and, on several headline variables including ten-metre and one-hundred-metre wind, competitive with or superior to the IFS control forecast at medium range. Energy traders have noticed. The operational fact that matters is not only accuracy at a fixed lead time. It is the combination of accuracy and update frequency. A model that is slightly better at noon and much fresher at 18:00 and 03:00 changes the information available at SIDC gates.

A second layer sits between the atmosphere and the cash register: the power model. One-hundred-metre wind is not farm output. Wake effects, icing, turbine availability, curtailment, hub-height mismatch and farm-specific power curves convert a good wind-speed forecast into a mediocre generation forecast, or a mediocre wind-speed forecast into a usable one. Deep-learning power models trained on reanalysis and farm data have been shown to improve multi-day generation forecasts relative to simple power-law conversions. In the laboratory below, a mundane histogram gradient-boosting model that takes an NWP-through-curve forecast and lagged realised generation as inputs captures a large share of the remaining error. That result is not a claim that gradient boosting replaces AIFS. It is a claim that the commercially relevant residual often lives in the second mapping.


### 2.5 Information, infrastructure and ordering

The political economy of infrastructure has long treated information as constitutive rather than ancillary. Electricity is an extreme case because it cannot be stored cheaply at system scale, because it is priced in sequential auctions with hard gates, and because its short-run supply function is convex. In such a market a private refinement of a public forecast is a private claim on the residual of a public good. Transparency regulation socialises the baseline. Foundation models and fine-tunes privatise the residual. The distributional question is who may occupy that residual: a trading house, a utility with SCADA, a TSO, a municipal co-operative, a vendor. The design question is whether fifteen-minute products, published ensembles, reserve-procurement rules and data-sharing obligations widen or narrow the private return to the residual. This paper does not resolve those questions. It supplies a mechanism and a set of magnitudes that make them concrete.


## III. INSTITUTIONAL ARCHITECTURE


### 3.1 Single Day-Ahead Coupling

Single Day-Ahead Coupling is the pan-European auction that jointly clears energy and cross-border transmission for the following day. Nominated electricity market operators collect bids and offers. An algorithm allocates energy and scarce interconnection so that, subject to network constraints, prices and net positions are consistent across bidding zones. The gate is around 12:00 Central European Time on $D{-}1$. At that moment tomorrow’s wind is not observed. Every wind bid is a forecast. Every thermal commitment that cannot be cheaply reversed is a forecast. Every storage schedule that is only partly flexible after noon is a forecast. The price that emerges is the shadow price of expected residual load given expected interconnection.

The move from hourly to fifteen-minute products changes the object being forecast. An hourly product can hide intra-hour ramps. A fifteen-minute product exposes them. Shape error becomes cash-flow. Models that update hourly, or that produce calibrated intra-hour trajectories, capture a rent that coarser models cannot. Market design here is again an information policy. It decides at what temporal resolution prediction may be sold.


### 3.2 Single Intraday Coupling

Single Intraday Coupling is the continuous cross-border market that remains open until shortly before delivery, supplemented by discrete intraday auctions. Its economic purpose is to allow agents to correct the day-ahead schedule as information arrives. For a wind-rich system the dominant flow of new information is meteorological. A downward revision of North Sea wind at 18:00 is an offer to buy in the German zone and, through coupling, a signal to hydro, gas and interconnectors elsewhere. Intraday prices can diverge sharply from the day-ahead outcome for the same delivery hour. A day that looked well supplied at noon can look tight by evening if the wind drops. The intraday price moves before any balancing action is taken. That is the market doing what it is designed to do. It is also the market in which a fresher AI run has its first cash value.


### 3.3 Balancing

After the last intraday gate the transmission system operator balances the system in real time using automatic and manual frequency restoration reserves and national products. The imbalance price settles the difference between each balance-responsible party’s final position and its actual injection or offtake. In many designs the imbalance price is more extreme than the day-ahead price and depends on the direction of the system imbalance. Systematic renewable forecast error is a primary source of that direction. A large offshore wind farm can, if it takes a large open position, move the imbalance price against itself. The laboratory below abstracts from that market-impact channel. The channel exists and is a further reason why prediction at scale is not a purely private statistical exercise.


### 3.4 Transparency as a public baseline

Regulation 543/2013 requires publication of a harmonised set of electricity market data. The ENTSO-E Transparency Platform is the resulting public good: day-ahead prices, actual and forecast load, actual generation by production type, wind and solar forecasts, installed capacity, physical flows, outages. Document types A.44, A.65, A.69 and A.75 are the series that a non-synthetic sequel to this paper would pull. The public baseline is not equal to the best private forecast. It is the common knowledge around which private refinements are built. Ignoring that distinction flattens the politics of prediction into a contest of algorithms.


## IV. THEORETICAL FRAMEWORK


### 4.1 Residual load and price

Let $D_t$ denote short-run demand, treated as approximately inelastic within the hour. Let $W_t$ be realised wind and $S_t$ realised solar. Residual load is

$$
R_t = D_t - W_t - S_t.
$$

Conventional plant is stacked by short-run marginal cost $c(\cdot)$, which is non-decreasing. Let $K_t$ be available dispatchable capacity. A compact representation of the clearing price is

$$
P_t = c(R_t) + \psi\left(\frac{R_t}{K_t}\right),
$$

where $\psi$ is a scarcity term that is negligible when residual load is well below available capacity and steep when residual load approaches or exceeds it. Differentiating with respect to wind gives a non-positive slope:

$$
\frac{\partial P_t}{\partial W_t} \le 0.
$$

The slope is small in well-supplied hours and large in tight hours. Convexity of $c$ and $\psi$ is the reason a mean-preserving spread in wind forecast error raises expected price impact and expected imbalance cost even if mean error is zero.


### 4.2 Forecasts as the day-ahead argument

The day-ahead auction does not see $R_t$. It sees an expected residual load constructed from a day-ahead load forecast and day-ahead renewable forecasts. Write $\hat{W}^{DA}_t$ for the wind forecast used at the day-ahead gate. Then

$$
P^{DA}_t = c\big(\hat{D}_t - \hat{W}^{DA}_t - \hat{S}^{DA}_t\big) + \psi(\cdot) + \varepsilon_t.
$$

Realised wind appears later, in the intraday and balancing stages. The quantity that must be re-traded is the forecast error. If the market internalises a better wind forecast, it clears a different stack. That is the counterfactual implemented in Section VII: the pricing rule is held fixed and only the wind input is replaced.


### 4.3 Imbalance cost

A merchant producer who bids the day-ahead point forecast and settles the residual at the imbalance price receives revenue

$$
\hat{W}\, P^{DA} + (W - \hat{W})\, P^{\mathrm{imb}}.
$$

Relative to a perfect-foresight day-ahead sale of realised wind at the day-ahead price, the imbalance cost is

$$
(W - \hat{W})\big(P^{\mathrm{imb}} - P^{DA}\big).
$$

The producer is punished when short in a short system and when long in a long system, that is, when the sign of own error agrees with the sign of system error. A better forecast reduces both the volume of imbalance and that covariance. In the laboratory the system error is identified with the fleet error, so the two channels move together. In a real system a small producer can be lucky or unlucky relative to the system. A large producer is the system.


### 4.4 Lead time as a product

Let $\sigma^2(\tau)$ be the variance of wind-speed forecast error at lead $\tau$. For classical NWP this variance declines in steps at official run times. For AI systems it can decline more continuously as new observations are assimilated and as new inferences are issued. The private value of a forecast is not only $\sigma^2$ at the day-ahead gate. It is the path of $\sigma^2$ from $D{-}1$ through $H{-}6$ to $H{-}1$, because each SIDC gate is a chance to trade the revision. A model that is only modestly better at noon and much better at $H{-}6$ can dominate a model that is excellent at noon and stale by evening. The lead-time ladder in the laboratory is a reduced-form representation of that product.


### 4.5 Ordering

Ordering has two meanings. The market is put in order: fewer last-minute scrambles, fewer wasted thermal starts, smaller reserve activations. Agents are ordered: ranked by the quality of their information set. A utility that fine-tunes a foundation weather model on its own supervisory data occupies a different rank from a municipal co-operative that bids the TSO’s public forecast. The first meaning is efficiency. The second is distribution. Both are produced by the same statistical improvement. Analyses that report only mean absolute error report neither.


## V. THE LABORATORY


### 5.1 Design principles

The laboratory is designed to be structural, inspectable and replaceable. Structural means that each series is generated from an explicit mechanism rather than from a generative adversarial network. Inspectable means that every intermediate object—wind speed, power fraction, load, residual, price, export, forecast error—can be plotted and summarised. Replaceable means that the economic layer does not depend on the generator. An ENTSO-E extract of day-ahead prices, load, wind forecasts and actual generation can be dropped in without rewriting the imbalance, commitment or counterfactual-price experiments.


### 5.2 Home zone

The home zone is calibrated to German and DE–LU regularities without claiming to be Germany. Mean load is 56.4 GW, with a weekday diurnal cycle, a weekend reduction and a winter elevation. Installed wind is 70 GW. The realised capacity factor in the two-year sample is 24.9 percent, higher in winter, which is consistent with a North Sea-influenced fleet. Solar capacity is 80 GW with a strong seasonal and diurnal envelope and a modest negative association with windy, cloudier hours. Residual load is load minus wind minus solar. It can be negative in high-renewable hours, which is the condition under which negative prices appear.


### 5.3 Power curve

The fleet power curve is an IEC-style piecewise function, smoothed at cut-out to reflect geographic dispersion. Cut-in is 3.5 metres per second. Rated speed is 12 metres per second. Cut-out is 25 metres per second. Between cut-in and rated speed, power rises with the cube of wind speed. Figure 1 plots the curve and its local sensitivity. Peak sensitivity is 17.6 GW per metre per second near 11.9 metres per second. That is the cubic band. A one-metre-per-second forecast error there is an entirely different economic object from a one-metre-per-second error at 15 metres per second, where the fleet is already rated. AI improvements concentrated in the cubic band are worth more than improvements in already-rated hours.

Figure 1. Smoothed fleet power curve (left) and local sensitivity of fleet output to wind speed (right). Peak sensitivity lies in the cubic band just below rated speed.


### 5.4 Prices and the neighbour

Home-zone price is a convex function of residual load: a hyperbolic tangent around a mid-stack level, plus piecewise linear terms that steepen above 22, 38 and 50 GW of residual load, plus a tightness term linked to a stylised gas short-run marginal cost that includes a heat rate and a carbon component, plus modest Gaussian noise. When residual load falls below 8 GW the function subtracts a negative-price correction. The resulting sample mean price is 48.57 EUR/MWh, with a right tail up to a few hundred euros and a left tail to −110 EUR/MWh. Negative-price hours are 19.5 percent of the sample. That share is high relative to earlier German years and not implausible relative to recent high-renewable springs, but it is a laboratory outcome, not an official frequency.

The neighbour is a hydro-rich, lower-price zone. Its residual is load minus a seasonal hydro process. The stylised net transfer capacity is 8,000 MW. Net export from the home zone is a clipped linear function of the price spread. Mean export in the sample is 187 MW. The interconnector is a shock absorber, not a complete equaliser. When the home zone is long on wind it tends to export; when it is short it tends to import. Section VII shocks expected home-zone wind and traces this rule.


### 5.5 Forecast errors

NWP and AI wind-speed errors are first-order autoregressive at the synoptic timescale, with a slight positive bias that is common in operational settings. The AI innovation variance is lower. Both errors are passed through the fleet curve, so megawatt error is state-dependent: the same wind-speed error produces more megawatt error in the cubic band. A lead-time family scales the same coloured noise to approximate $D{-}1$, $H{-}6$ and $H{-}1$. Persistence is intentional. If this afternoon’s forecast is high, tonight’s is probably high as well. SIDC revisions therefore arrive in clusters. A better model is valuable pathwise, not only on average.

## VI. EMPIRICAL DESIGN


### 6.1 Diagnostics

Four reduced-form objects organise the economic geography of wind in the laboratory. The first is the correlation of price with wind, solar, load and residual load. The second is the binned merit-order curve obtained by splitting hours into wind octiles. The third is the value factor of wind and of solar. The fourth is a Dunkelflaute indicator: winter months, wind capacity factor below eight percent, solar below eight percent of capacity, and load above its median. These objects do not prove causality. They establish that the laboratory reproduces the qualitative facts on which the thesis depends: a negative wind–price relationship, cannibalisation, and a scarce tail of dark-still winter hours.


### 6.2 Electricity price models

Histogram gradient-boosting models are trained on the first year and scored on the second. Features are a noisy day-ahead load forecast, calendar harmonics, price lags at twenty-four and one hundred and sixty-eight hours, solar, and either the NWP or the AI day-ahead wind forecast. The operational experiment keeps the NWP-trained model and replaces only its wind feature with the AI forecast. The resulting price path is the market that would have cleared had the mapping from fundamentals to price remained the same while the wind number improved. Mean absolute gap, mean signed gap and the ninetieth percentile of the absolute gap are reported. A ridge regression on standardised features is reported so that the sign and rank of the wind coefficient can be read without opening a tree ensemble.


### 6.3 Quantile wind

Linear quantile regressions at the tenth, fiftieth and ninetieth percentiles of realised wind on the NWP point forecast and calendar features are estimated on year one and scored on year two. Empirical coverage of the ten-to-ninety interval is compared with the nominal eighty percent. Mean interval width is reported in gigawatts. The model is intentionally linear and small. Operational ensembles are not. The point is to make tail information visible as an object that a unit-commitment rule could use.


### 6.4 Merchant cash-flow

A one-hundred-megawatt slice of the fleet bids the day-ahead point forecast and settles the residual at an imbalance price equal to the day-ahead price plus a mark-up that grows with signed system wind error to the power 0.85, plus noise. Figures are annualised by the ratio of hours in a year to hours in the sample. The slice size is chosen so that results can be compared in spirit with published industry evaluations that report imbalance-cost reductions per one hundred megawatts. The imbalance function is reduced-form. Real automatic frequency restoration reserve merit orders are stepwise. Results are therefore orders of magnitude.


### 6.5 Thermal commitment

A stylised five-hundred-megawatt combined-cycle unit is committed day-ahead if expected residual load exceeds 40 GW. Expected residual load is load minus the day-ahead wind forecast minus solar. Starts are off-to-on transitions. False commitments are hours in which the unit is committed although realised residual load would not have required it. The rule is brutal. It ignores start costs as an explicit objective, minimum up and down times, and the option to wait for the intraday. It is nevertheless sufficient to show that forecast error purchases thermal motion that realisation would not have required.


### 6.6 Power model and cross-border shock

A histogram gradient-boosting power model is trained to predict realised wind from the NWP-through-curve forecast, lags at one, twenty-four and forty-eight hours, and calendar harmonics. Test-year mean absolute error is compared with the raw NWP curve. Separately, home-zone day-ahead wind in the test year is shocked by minus eight, minus four, zero, plus four and plus eight gigawatts inside the NWP price model. Mean reconstructed home-zone price and mean stylised export are recorded. The exercise is not flow-based market coupling. It is a reduced-form reminder that a German wind number is an input to neighbouring dispatch.


## VII. RESULTS


### 7.1 Sample structure

The laboratory comprises 17,520 hourly observations. Mean wind speed is 7.38 metres per second. Mean wind generation is 17.5 GW. Mean solar generation is 8.0 GW. Mean load is 56.4 GW. Mean residual load is 30.9 GW. Mean home-zone price is 48.57 EUR/MWh. Mean neighbour price is 25.15 EUR/MWh. Mean export is 187 MW. The wind capacity factor is 24.9 percent. These levels are inside the range of published German-scale regularities without being a reconstruction of any particular year.

### 7.2 Merit order, value factors and Dunkelflaute

Price correlates with wind at −0.603, with solar at −0.598, with load at +0.600 and with residual load at +0.893. The signs are those that merit-order theory requires. The wind value factor is 0.400. The solar value factor is −0.331. Solar’s negative value factor in this sample reflects the coincidence of high solar hours with already-low residual load and with the laboratory’s negative-price correction. The qualitative ranking—solar more cannibalised than wind—matches the European value-factor literature. Dunkelflaute-like hours are 4.0 percent of the sample. Their mean price is 196.1 EUR/MWh against an all-hours mean of 48.6 EUR/MWh. The tail is where convexity lives. It is also where a two-gigawatt wind error ceases to be a balancing nuisance and becomes a scarcity event.


### 7.3 Forecast accuracy by lead time

Table I reports wind generation forecast error. Day-ahead NWP mean absolute error is 9,078 MW, or 12.97 percent of installed capacity. Day-ahead AI mean absolute error is 5,971 MW, or 8.53 percent of capacity. The reduction is 34.2 percent. At $H{-}6$ the figures are 4,984 MW and 2,493 MW. At $H{-}1$ they are 1,968 MW and 827 MW. Two comments follow. First, both regimes improve as delivery approaches, which is the classical structure of weather error. Second, the AI advantage is present at every rung of the ladder. The commercially relevant object is the whole ladder, because each rung corresponds to a market gate.

### 7.4 Electricity prices as an information experiment

Table II reports out-of-sample electricity price forecast scores. The model that sees NWP wind attains a mean absolute error of 21.77 EUR/MWh and an R² of 0.772. The model that sees AI wind attains 19.08 EUR/MWh and 0.830. The improvement is expected: a better fundamental should help a fundamentals-based price model. The more important experiment holds the NWP pricing rule fixed and replaces only the wind input. Mean absolute price shift is 12.90 EUR/MWh. Mean signed shift is 0.34 EUR/MWh. The ninetieth percentile of the absolute shift is 30.89 EUR/MWh. The signed mean is near zero because AI is not systematically high or low in a way that always raises or lowers the stack. The absolute mean and the upper percentile are large because the hours in which the two wind numbers disagree are often hours in which the supply curve is steep. That is reordering. It is not a statement that average prices fall. It is a statement that the path of prices, and therefore the path of dispatch, changes.

Table II. Out-of-sample day-ahead electricity price forecast scores

Ridge coefficients on standardised features in the NWP price model rank day-ahead load first with a positive sign and NWP wind second with a negative sign. Solar is also negative. Lagged prices are positive, as persistence requires. Calendar harmonics are smaller. The linear skeleton of the laboratory is therefore the skeleton of a merit-order market: more expected demand raises price; more expected wind lowers it. Tree ensembles capture convexity around that skeleton. The counterfactual experiment asks what happens when the skeleton is fed a different wind number. Table II and the shift statistics are the answer.


### 7.5 Probabilistic wind

Tenth, fiftieth and ninetieth quantile models of realised wind, using the NWP point forecast and calendar features, produce a ten-to-ninety interval with 82.9 percent empirical coverage on the test year against a nominal 80 percent. Mean width is 19.66 GW. Coverage close to nominal is a limited achievement for a linear model on a structural simulator; it would be more informative on ENTSO-E data. The economic use is nevertheless clear. A thermal owner who treats the ninetieth percentile of wind as the optimistic renewable case, and who starts plant only when even that case leaves residual load high, will start less often in vain than an owner who keys off the median. In a convex system the product being sold is that tail, not the median.


### 7.6 Merchant imbalance

Table III reports annualised cash-flow for a one-hundred-megawatt slice. Imbalance cost is 6.19 million EUR under NWP and 2.86 million EUR under AI. The difference is 3.33 million EUR per year for that slice, under this imbalance-price function. Net revenue flips from negative to positive because, in the NWP world, wrong-side exposure in stressed hours overwhelms the day-ahead capture; in the AI world it does not. These levels are larger than some published real-market evaluations for comparable capacity. The laboratory imbalance function is intentionally stressful, and the slice inherits the system error one-for-one. The qualitative result is what the thesis requires: a tighter forecast reduces both the volume of imbalance and its coincidence with expensive hours, and the cash difference is first-order relative to the energy being sold.

### 7.7 Thermal starts

Table IV reports the commitment experiment. Realised residual load would have required 645 starts and 7,346 committed hours under the forty-gigawatt rule. The NWP residual forecast produces 874 starts and 6,787 committed hours, with 1,224 hours in which the unit is committed although realisation would not have required it. The AI residual forecast produces 782 starts and 7,255 committed hours, with 970 false hours. AI does not eliminate wasted motion. It reduces it. The reduction is the thermal bill of forecast error, paid by owners and, through offer prices, by the system. It does not appear on the wind farm’s invoice. It is part of the ordering that prediction performs.

### 7.8 Power model

The supervised power model that sits on the NWP-through-curve forecast reduces test-year mean absolute error from 8.995 GW to 2.498 GW, a 72.2 percent reduction. The figure is large because lagged realised wind is a powerful feature in an autoregressive weather system and because the learner can correct state-dependent curve error. In a real stack the same architecture would be trained on farm SCADA rather than on the simulator’s own lag. The methodological point stands. Weather models and power models are different products. A market that publishes a public weather run still leaves a private power-model residual. That residual is one of the places where prediction-as-power is actually sold.


### 7.9 Cross-border shock

Table V reports the cross-border experiment. Starting from the test-year NWP wind path, an eight-gigawatt downward shock raises mean reconstructed home-zone price from 46.7 to 58.8 EUR/MWh and mean export from 172 to 269 MW. An eight-gigawatt upward shock lowers mean price to 36.6 EUR/MWh and mean export to 92 MW. The stylised interconnector does not bind at the ninety-eight percent threshold in this sample, so the adjustment is interior. The sign pattern is the coupled-market pattern: less expected home-zone wind, higher home-zone price, more export-or-less-import from the neighbour’s perspective depending on sign convention. A private refinement of the home-zone wind number is therefore an input to neighbouring hydro. Prediction orders assets that are not in the same country as the forecast error.

### 7.10 Reading the magnitudes together

No single table is the result. The result is the joint pattern. AI reduces error at every lead time. The hours in which the two forecasts disagree are hours in which reconstructed prices move by amounts that are large relative to mean prices. Merchant imbalance cost falls by a first-order amount for a small slice. Thermal plant is committed falsely less often. Neighbour export moves when expected home-zone wind moves. Value factors remain below one, so cannibalisation is not repealed. Dunkelflaute hours remain expensive, so convexity is not repealed. What changes is who is told, in time, that those hours are coming, and who is still surprised.


## VIII. WHO IS ORDERED, AND BY WHOM


### 8.1 Agents

A wind producer with access to the finer forecast faces a smaller imbalance volume and less wrong-side exposure. If the producer also has flexible volume—curtailment, a co-located battery, a hybrid offtake—it can move that volume into hours the forecast marks as scarce. A producer without the finer forecast sells a coarser description of the same weather and buys back the difference at a stressed price.

A combined-cycle or coal owner faces fewer wasted starts when day-ahead wind was too low and fewer emergency ramps when it was too high. Start costs are the hidden invoice of bad wind forecasts. They are social in the sense that they enter offer prices and reserve needs even when they are privately incurred. A better public forecast would socialise the saving. A better private forecast privatises it for those who see the revision in time to cancel the start.

Hydro owners and interconnectors absorb a smaller share of one zone’s surprise if that zone’s day-ahead schedule is closer to realisation. They still absorb the residual. Coupled Europe does not disappear because one zone buys a better model. It changes the amplitude of the shock that coupling is asked to absorb. Storage and demand response can key charge and discharge decisions off a sharper residual-load path rather than off a surprise. That option is valuable only to agents who see the path.

A transmission system operator activates less automatic and manual reserve if the day-ahead schedule is closer to realisation. Reserve procurement itself becomes a forecast product. Consumers may see lower balancing-energy cost in aggregate. They do not automatically see a lower retail tariff. The incidence depends on how imbalance cost, reserve cost and wholesale prices pass through contracts that are themselves often forecast-based.

A forecast vendor sells a private refinement of a public TSO or ECMWF baseline. The product is the residual. The client list of such vendors is a map of who is allowed to occupy the finer filtration. That map is part of the ordering.


### 8.2 Design instruments

Fifteen-minute products raise the return to intra-hour shape prediction. Discrete intraday auctions create additional common gates at which a new run can be capitalised by everyone who sees it, or only by those who subscribe, depending on who receives the run. Published ensembles socialise tail information that point forecasts hide. Mandatory sharing of certain meteorological inputs or of anonymised farm production can narrow the private residual. Reserve-procurement rules that score providers on forecast quality treat prediction as a system service rather than as a private edge. None of these instruments is neutral. Each changes the private return to AI. Treating them as purely technical is a political choice.


### 8.3 What this is not

This is not a claim that public forecasts should be abolished. The public baseline is what makes a market common knowledge rather than a set of private rumours. This is not a claim that private models should be banned. Fine-tuning on farm data uses information that a central model cannot see. This is not a claim that AI weather models are already uniformly better than IFS in every variable at every lead time. They are better in some, competitive in others, and operationally different in update frequency. The claim is narrower and sharper. In a market that clears forecasts, a systematic advantage in the residual of the public forecast is an advantage in who runs. Calling that advantage prediction-as-power is a way of refusing to treat it as weather.


## IX. LIMITATIONS AND A NON-SYNTHETIC SEQUEL

The laboratory is an argument with numbers, not an official impact assessment. Series are synthetic. Coefficients are orders of magnitude. Flow-based market coupling, bidding-zone splits inside Italy and the Nordics, and unit-commitment integer constraints are absent or reduced-form. They matter for both price formation and cross-border absorption. The imbalance-price function is smooth. Real automatic frequency restoration reserve curves are stepwise and sometimes invert. No agent in the laboratory is large enough to move the imbalance price with its own bid. The offshore-farm market-impact problem studied in recent work is therefore left out. Solar, outages, fuel-price dynamics, carbon-price dynamics and curtailment are stylised. Quantile regressions are linear. Operational ensembles are not. The supervised power model has access to lags that a true day-ahead gate would only partly have, depending on publication lags.

A production replication is well defined. Pull ENTSO-E document type A.44 for day-ahead prices, A.65 for load, A.69 for wind and solar forecasts, and A.75 for actual generation by type, for DE–LU, DK1, DK2, SE3 and SE4, at hourly or fifteen-minute resolution. Score ECMWF IFS against AIFS on one-hundred-metre wind at farm or reanalysis coordinates. Rebuild the lead-time ladder from issued-run archives. Re-estimate the price models, the imbalance cash-flow using published imbalance prices, and the cross-border flows using A.11 physical flows. Compare value factors on realised series. The code structure of the companion notebook is arranged so that those extracts replace the generator without rewriting the economic layer. Until that is done, the tables in Section VII should be cited as laboratory magnitudes.

A further sequel would treat market impact. A large offshore fleet that bids a private forecast which the rest of the market does not share can move intraday and imbalance prices. The rent to prediction then includes a pecuniary externality. Optimal bidding under that externality is a different paper. It is mentioned here because it is the point at which prediction-as-power meets classical market power and can no longer be separated from it by a clean institutional line.


## X. CONCLUSION

In a wind-rich, coupled European market the day-ahead auction is a forecast-clearing institution. The object that is priced is residual load as expected at gate closure. Wind’s position at the foot of the merit order amplifies forecast revisions into price revisions. The mapping is convex. The same gigawatt error is cheap in a well-supplied hour and expensive in a tight or Dunkelflaute hour. That is why tail forecasts, not only mean absolute error, are the commercial product.

Artificial-intelligence weather and power models reduce error at every lead time on the ladder from the day-ahead gate to the last hour. In this laboratory the reduction is 34.2 percent of day-ahead wind mean absolute error. It appears as a 12.90 EUR/MWh mean absolute shift in reconstructed day-ahead prices, a 3.33 million EUR annualised imbalance-cost reduction for a one-hundred-megawatt slice, fewer false thermal commitments, a usable probabilistic interval around the public-style point forecast, a power-model residual that a supervised learner can still capture, and a different interconnector schedule when expected wind is shocked. That pattern is a reallocation of who runs, not only a statistical improvement.

The rent to prediction is appropriable. Public TSO and ECMWF forecasts socialise a baseline. Proprietary models capture the residual. Market design that ignores this residual treats an information asymmetry as if it were weather. Prediction is power because, in this market, prediction is what the power system is told to become.

## APPENDIX D. EXTENDED INSTITUTIONAL NOTES

This appendix records institutional detail that would interrupt the argument in the main text but that a reader implementing a sequel will need. Bidding zones are not countries. DE–LU has been a common zone since late 2018. Austria is separate. Denmark is split into DK1 and DK2. Sweden is split into four zones. Norway is split into several. Italy is split into multiple mainland and island zones. Price formation inside a zone assumes sufficient internal transmission; congestion inside a zone is managed by the TSO rather than by the day-ahead algorithm. Flow-based market coupling in the Core region allocates a set of critical network elements rather than a simple net transfer capacity on each border. The laboratory’s clipped linear export rule is closer to an NTC world than to a flow-based world. Replacing it with a flow-based engine would change Table V and would not change Tables I–IV.

Balance-responsible parties are the legal persons who hold positions in the markets and who are settled at the imbalance price. A wind farm may be its own balance-responsible party or may sit inside a larger portfolio. Portfolio netting changes the private value of a forecast. A utility that is long wind and long flexible demand can offset errors internally before they become imbalance. A stand-alone merchant farm cannot. Prediction-as-power is therefore stronger, privately, for agents who cannot net and still strong, socially, for agents who can, because social imbalance is the sum of what is not netted.

Reserve products are themselves forecast objects. Dimensioning of automatic frequency restoration reserve often uses distributions of imbalance that are, in large part, distributions of renewable forecast error. A structural improvement in day-ahead and intraday renewable forecasts should, over time, reduce reserve needs. The saving is slow to appear because dimensioning rules are conservative and because other sources of imbalance remain. It is nevertheless part of the social return to prediction and is not captured in Table III.

Negative prices deserve a longer note. They occur when must-run generation plus inflexible renewable output exceeds demand net of export and storage offtake. They are a feature of a market that allows the stack to clear below zero rather than curtailing by administrative fiat. They are also a signal that the day-ahead description of renewable output was high, or that flexibility was not positioned, or both. A better day-ahead wind forecast does not abolish negatives. It announces them. Storage, demand response and interconnectors can then be on the correct side of the announcement. Surprise negatives are more expensive than announced negatives because the flexibility that could have absorbed them is pointing the wrong way.

Dunkelflaute deserves a matching note. The word names a stretch of still, dark weather in winter. In the laboratory it is operationalised as a set of hourly flags rather than as a multi-day event study. Multi-day events are the operational nightmare because they exhaust storage, strain interconnectors and force thermal plant that was not in the money yesterday to start today. Forecast quality at the three-to-ten-day horizon, which is where AI weather models have been particularly advertised, matters more for those events than day-ahead mean absolute error. The laboratory’s lead-time ladder stops at $D{-}1$. Extending it is a natural sequel.


## APPENDIX E. METHOD NOTES FOR REPLICATION

Random numbers use a single NumPy Generator seeded at 42. The autoregressive weather process is generated in a Python loop rather than in closed form so that the persistence parameter is explicit. The power curve is applied pointwise. Prices are computed from residual load without iteration. There is no equilibrium loop in which price feeds back into demand. Elastic demand would dampen the merit-order effect and would reduce the value of forecast improvements that operate through price convexity. It would not remove the imbalance channel.

Training and test splits are chronological. The cut is one year after the first timestamp. No k-fold shuffle is used. Electricity series are not exchangeable across time. Leakage from future hours into past hours would flatter every score in Section VII. Lagged prices at twenty-four and one hundred and sixty-eight hours are available at the day-ahead gate for the corresponding delivery hour of the previous day and week. The noisy load forecast is a stand-in for a TSO day-ahead load forecast and is not itself an AI object in this paper.

Mean absolute error is reported in megawatts and as a percentage of installed capacity. Root-mean-square error is reported where it changes the ranking, which it does not in Table I. Pinball loss is the proper score for the quantile models and is not reported because the linear quantile models are illustrative. A sequel should report pinball loss and interval width jointly.

Figures are drawn with a serif typeface at 140 dots per inch. They are laboratory figures. They should not be reproduced as if they were plots of ENTSO-E series. The companion notebook regenerates them from the same seed.

## APPENDIX H. WORKED INTERPRETATION OF THE MAIN MAGNITUDES

A 12.90 EUR/MWh mean absolute price shift on 56 GW of hourly demand is a transfer of about 722,000 EUR in that hour between producers as a class and consumers as a class, ignoring hedges. The ninetieth percentile of 30.89 EUR/MWh on the same energy is about 1.73 million EUR in that hour. These figures are scale checks, not welfare estimates. They show that an information upgrade that looks modest in a forecast table is first-order once it is multiplied by tens of thousands of megawatts.

One hundred megawatts of wind at a 24.9 percent capacity factor produces about 218 GWh per year. An imbalance-cost reduction of 3.33 million EUR on that energy is about 15 EUR/MWh of energy produced. The laboratory imbalance function is stressful by design. A sequel that uses published German imbalance prices should expect a smaller euro figure and the same sign.

Two hundred and fifty-four avoided false committed hours on a 500 MW unit are 127 GWh of plant-hours not committed in vain. If a start costs tens of thousands of euros in fuel and wear, the annual saving is not a rounding error. That bill does not appear on the wind farm’s invoice.


## APPENDIX I. THE DAY IN THE LIFE OF A FORECAST

At 08:00 on $D{-}1$ a utility updates tomorrow’s wind using the overnight NWP run and any later AI inference. At 11:00 bids are assembled. At 12:00 SDAC closes. The utility has sold a volume equal to its day-ahead point forecast or a transformation of it. After publication, prices embed whatever wind number the mass of the market used.

At 15:00 a new AI run arrives. Expected own-fleet wind for a peak hour has fallen by 80 MW and expected system wind by more than 1 GW. The utility buys 80 MW in SIDC. Others who see a similar revision do the same. The intraday price rises. Neighbour hydro offers more water through coupling. A combined-cycle unit that was off at the day-ahead stage receives a start instruction. At 03:00 on day D a further run arrives. The remaining revision is small and is traded. At delivery the last 15 MW is imbalance. The premium is paid on 15 MW rather than on the 80 MW that would have remained open without the afternoon trade. Every step is a sale of prediction. An agent who owns only the public noon forecast occupies only the first step with confidence.


## APPENDIX J. WHY MEAN ABSOLUTE ERROR IS NOT ENOUGH

Two forecasts can share a mean absolute error and differ in the hours in which they err. Error in a Dunkelflaute hour is not the same object as error in a saturated-wind hour. Pinball loss at extreme quantiles, slope-weighted error, and imbalance cost are better economic scores. Table I reports MAE because weather models are still compared in that language. Table III exists because markets do not pay in MAE.

The cubic band on the power curve is the physical counterpart. Error in metres per second is not error in megawatts. Error in megawatts is not error in euros. Each mapping is nonlinear. A paper that stops at metres per second has not reached the market.


## APPENDIX K. PUBLIC BASELINES AND PRIVATE RESIDUALS

The public baseline is a bundle: the TSO wind forecast, the IFS run, the published load forecast, outages, and the published transfer domain. Private residuals are also a bundle: a finer inference, a farm power model, availability, faster ingest. Prediction-as-power lives in the residual bundle. Publishing one more official point forecast narrows one residual and may widen another as effort reallocates.

Transparency of outcomes makes a market auditable. Transparency of inputs can narrow rents. Transparency of model weights is a harder demand. Foundation models trained on public reanalysis still produce private inference products because inference is costly and because SCADA fine-tunes are not public. A regulator who wants to socialise the residual must name the layer.


## APPENDIX L. RELATION TO CLASSICAL MARKET POWER

Classical market power is the ability to raise price by withholding capacity or offering above short-run marginal cost in a concentrated residual market. Prediction-as-power requires concentration in the information set, not in plant. A trading house with no plant and a superior model has the first without the second. A municipal generator with plant and only the public forecast has the opposite. When they coincide, a high thermal offer on a day of privately expected low wind may be a legitimate residual-load forecast or withholding dressed as a forecast. Published ensembles give the rest of the market a benchmark.

If a large farm’s private forecast is not shared, its trades can move prices. Optimal bidding then includes a concealment motive. That problem is named, not solved, here. Pinson, Chevallier and Kariniotakis [14] treat imbalance-aware bidding for a price-taking producer. They do not claim that the producer sets the imbalance price.


## APPENDIX M. THE COUNTERFACTUAL, SLOWLY

Year one estimates a mapping from features to day-ahead price, including the NWP wind forecast. Year two is held out. The mapping is applied twice: once with NWP wind and once with AI wind in the same slot. Everything else is identical. The difference is therefore the wind input. The mapping is a tree ensemble, so disagreement is not a constant. Hours of large meteorological disagreement are often hours of unusual residual load. Convexity amplifies that into a price disagreement whose ninetieth percentile is more than twice the mean absolute shift.

A long-run market would re-estimate the mapping. The experiment measures the immediate reordering when an existing rule is fed a new number. That is what happens at noon on the day a new forecast feed is switched on.


## APPENDIX N. VALUE FACTORS

The laboratory wind value factor of 0.400 and solar value factor of −0.331 are severe relative to many published annual German figures. Hirth [8], [30] documents value-factor decline with penetration and a stronger solar penalty at lower shares. The ranking—solar more cannibalised than wind—should travel. The laboratory levels should be re-estimated on ENTSO-E series before they enter a policy sentence about captured prices.


## APPENDIX O. FLEXIBILITY

Prediction without flexibility is an invoice. Flexibility without prediction is a guess. Batteries, demand response and hydro that cannot see a sharp residual-load path still act, but on a coarser signal. Hardware subsidies without a better public residual-load description buy machines that cannot see. Published ensembles without storage and demand response publish numbers that cannot act. Unit commitment is flexibility with a fixed cost. Table IV is the cost of taking that decision on a worse forecast.


## APPENDIX P. GEOGRAPHY

The home zone is Germany-like because Germany combines a large wind fleet, nuclear exit, and a central position in coupling. The mechanism is not German property. Denmark is more wind-dominated and more tied to Nordic hydro. Iberia is wind-and-solar with limited export to France. Ireland is small and exposed to a single weather object. Italy’s multiple zones make intra-national prediction an inter-zonal problem. In each case the day-ahead auction clears a description of weather. Coupling exports the consequences of a description that may not have been shared.

## APPENDIX V. CLOSING

A market turns descriptions into allocations. When the description is weather and the allocation is which machines run, forecast quality is part of the allocation. Artificial intelligence intensifies that fact by increasing the number of gates at which a private residual can be sold. Improving the public baseline shrinks the residual. Leaving the public baseline stale inflates it. That is a design choice even when no one names it as one.

Prediction is power because, in this market, prediction is what the power system is told to become. The sentence does not require a particular laboratory cell. It requires the institutional facts of SDAC, SIDC and balancing [1], [3], the convex residual supply curve documented in the merit-order literature [6], [7], and the empirical fact that AI and NWP descriptions of wind are not the same description [4], [5], [11], [33].
